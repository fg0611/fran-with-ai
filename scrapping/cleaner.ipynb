```python
import numpy as np
import pandas as pd
```

```python
ruta_base = './archivos/'
archivo = 'respuestos_ford.csv'
df = pd.read_csv(ruta_base+archivo)
df.head()
```

```python
elim = [0, 2]
df = df.drop(df.columns[elim], axis=1)

# df = df.iloc[:,1:]
# df2 = df[1]
# df2.head()
df
```

```python
# 1. Verificar valores nulos
nulos = df.isnull()
# 2. Verificar longitud <4 para no nulos
cortos = df.apply(lambda col: col.dropna().astype(str).str.len() < 5)
# Combinar condiciones
df['es_invalida'] = (nulos.all(axis=1) | cortos.all(axis=1))
df
```

```python
condicion = df.iloc[:,2].isnull() | (df.iloc[:,2].astype(str).str.len() < 10) | df.iloc[:,-1].astype(bool)
df = df[~condicion]
```
