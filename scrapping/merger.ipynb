```python
import pandas as pd
import os
import re
```

```python
# 1. Configuración inicial
input_folder = './mod/'  # Cambia esto
output_file = 'repuestos_merged.csv'
columnas_relevantes = ['nombre', 'dir', 'tlf']  # Ajusta con tus nombres reales
```

```python

def solo_digitos(tlf):
   return re.sub(r'\D', '', str(tlf))

# 2. Procesar todos los archivos CSV
dfs = []
for filename in os.listdir(input_folder):
    if filename.endswith('.csv'):
        filepath = os.path.join(input_folder, filename)
        df = pd.read_csv(filepath)
        dfs.append(df)

# 3. Consolidación y eliminación de duplicados
if dfs:
    consolidated = pd.concat(dfs, ignore_index=True)
    consolidated = consolidated.iloc[:, :-2]
    consolidated.iloc[:,2] = consolidated.iloc[:,2].str.replace(r'\D+', '', regex=True)

        # Verificación antes de eliminar duplicados
    print(f"\nRegistros totales antes de eliminar duplicados: {len(consolidated)}")
    print(f"Duplicados detectados: {consolidated.duplicated().sum()}")
    # Eliminar duplicados considerando todas las columnas
    consolidated = consolidated.drop_duplicates()
    print(f"Registros totales: {len(consolidated)}")
    
    # Opcional: eliminar duplicados basados en columnas clave
    # consolidated = consolidated.drop_duplicates(subset=['col1', 'col2'])
    
    # Guardar resultado
    consolidated.to_csv(output_file, index=False)
    print(f"Archivo consolidado guardado como {output_file}")
else:
    print("No se encontraron archivos CSV en la carpeta")
```
