```python
import pandas as pd
import numpy as np
```

```python
ruta_base = 'archivos/'
nombre_archivo = 'const_merged' 
origen = f'{ruta_base}{nombre_archivo}.csv'
# print(origen)
df = pd.read_csv(origen)
X = 5  # Número de partes que quieres

dfs = np.array_split(df, X)
```

```python
# dfs[0]
# guardarlos en archivos
if dfs:
    for idx, section in enumerate(dfs):
        section.to_csv(f'{ruta_base}{nombre_archivo}_{idx}.csv', index=None)
```
