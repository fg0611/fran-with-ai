```python
import numpy as np
import pandas as pd
import re
```

```python
ruta_base = './archivos/'
sub = 'busquedas/'
query = 'const/'
archivo = 'ab.csv'
df = pd.read_csv(ruta_base+sub+query+archivo)
df.head()
```

```python
# eliminar datos donde no hayan telefonos
df = df.dropna(subset=['UsdlK'])
df.head(2)
```

```python
elim = [0, 2]
df = df.drop(df.columns[elim], axis=1)

# df = df.iloc[:,1:]
# df2 = df[1]
# df2.head()
df
```

```python
# 1. Verificar valores nulos
nulos = df.isnull()
# 2. Verificar longitud <4 para no nulos
cortos = df.apply(lambda col: col.dropna().astype(str).str.len() < 5)
# Combinar condiciones
df['es_invalida'] = (nulos.all(axis=1) | cortos.all(axis=1))
df
```

```python
condicion = df.iloc[:,2].isnull() | (df.iloc[:,2].astype(str).str.len() < 10) | df.iloc[:,-1].astype(bool)
df = df[~condicion]
```

```python
# arreglar numero de telefono
def arreglar_numero (num):
    return re.sub(r'\D', '', str(num))
# [df.iloc[:, 2]]
df.iloc[:, 2] = df.iloc[:, 2].apply(lambda x: arreglar_numero(x))
```

```python
# Drop column 'ah5Ghc 3' if it exists
df = df.drop(columns=['ah5Ghc 3', 'FQ2IWe src', 'e4rVHe'], axis=1, errors='ignore')
df.head()
```
