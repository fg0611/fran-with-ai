```python
import pandas as pd
import numpy as np
import re
```

```python
ruta_base = './archivos/'
sub = 'excels/'
archivo = 'db1.xlsx'
final = ruta_base+sub+archivo
# print(final)
df = pd.read_excel(final)
df.columns = [name.lower() for name in df.columns]
df.head(2)
```

```python
# arreglar numero de telefono
def es_telefono (num):
    if pd.isna(num) or not isinstance(num, (str, int, float)):
        return None
    if isinstance(num, float): # Convert float to integer first to remove decimals, then to string
        num = int(num)
    tlf = re.sub(r'\D', '', str(num))
    if len(tlf) in [10, 12, 14]:
        return tlf
    return None

def buscar_telefono(fila):
    for valor in fila:
        valor = es_telefono(valor)
        if valor:
            return valor
```

```python
df['tlf'] = df.apply(buscar_telefono, axis=1)
df = df.loc[:, ["cliente", "edad", "tlf"]]
df['cliente'] = df['cliente'].str.title()
df.columns = ['nombre', 'edades', 'tlf']
filtro_sin_tlf = df['tlf'].notna()
df = df[filtro_sin_tlf]
```

```python
df.to_csv('db2.csv', index=False)
```
